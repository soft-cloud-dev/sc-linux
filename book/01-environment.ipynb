# 1 · Prepare the build host

## Goal

Confirm that the host can build a **Debian 13 amd64** RAW image and boot-test it with QEMU. This chapter changes nothing.

## Prerequisites

Use a Linux x86-64 host with Docker, Git, Make, Python 3.11+, OpenSSL and at least 40 GiB of available space. Docker Desktop is not qualified for this workflow.

```python
from pathlib import Path
import os, subprocess, sys
START = Path.cwd().resolve()
ROOT = next((p for p in (START, *START.parents) if (p/"myst.yml").is_file() and (p/"scripts/build.sh").is_file()), None)
assert ROOT is not None, "Run from the SC Linux checkout (make book-lab)"
def run(*argv):
    print("+", " ".join(map(str, argv)), flush=True)
    return subprocess.run([str(a) for a in argv], cwd=ROOT, check=True)
print("Source:", ROOT)
```

## Action 1 · Identify the platform

```python
import platform, shutil
print("OS:", platform.system(), "Architecture:", platform.machine())
assert platform.system()=="Linux", "Use an amd64 Linux host"
assert platform.machine() in ("x86_64","AMD64"), "The current image target is amd64"
for binary in ("docker","git","make","python3","openssl"):
    print(binary, shutil.which(binary) or "MISSING")
    assert shutil.which(binary), f"Install {binary}"
```

## Action 2 · Check storage and Docker

```python
disk=shutil.disk_usage(ROOT)
print(f"Free disk: {disk.free/2**30:.1f} GiB")
assert disk.free >= 40*2**30, "Reserve at least 40 GiB"
run("docker","info","--format","{{.OSType}} {{.Architecture}}")
```

## Action 3 · Inspect Git state

Image builds require every tracked change to be committed; artifacts belong under ignored `build/`.

```python
run("git","status","--short")
run("git","rev-parse","--short","HEAD")
```
