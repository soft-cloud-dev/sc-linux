# 1. Validate SC Linux

Run from the repository root. This notebook verifies configuration, test cases, and shell syntax. It does not require Docker or a FreeIPA server.

```python
from pathlib import Path
import subprocess
ROOT = Path.cwd().resolve()
assert (ROOT / "Makefile").is_file(), "Start Jupyter from the repository root"
def run(*args):
    print("+", " ".join(map(str, args)), flush=True)
    subprocess.run([str(arg) for arg in args], cwd=ROOT, check=True)
```

## Check repository sources

Any failure stops the notebook and displays the failing command.

```python
run('make', 'check')
```
