# 2. Build SC Linux

Produces the Debian 13 amd64 RAW disk image and UEFI installer ISO.

**Prerequisites:** amd64 Linux, Docker, Git, OpenSSL, Make, 40 GiB free space, verified IPA CA certificate and fingerprint in `trust/`, and a clean committed Git tree. The Docker build requires privileged access.

```python
from pathlib import Path
import subprocess
ROOT = Path.cwd().resolve()
assert (ROOT / "Makefile").is_file(), "Start Jupyter from the repository root"
def run(*args):
    print("+", " ".join(map(str, args)), flush=True)
    subprocess.run([str(arg) for arg in args], cwd=ROOT, check=True)
```

## Inspect inputs

```python
import json
for name in ("config/build.json", "config/identity.json"):
    print(name)
    print(json.dumps(json.loads((ROOT/name).read_text()), indent=2))
run('git','status','--short')
```

## Verify the CA and configuration

```python
run('python3','scripts/check.py','--require-trust')
```

## Build the pinned toolchain

```python
run('make','builder')
```

## Build the RAW image and installer ISO

```python
run('make','images')
```

## Inspect generated files

```python
output=ROOT/'build/output'
for name in ('sc-linux.raw.zst','sc-linux.iso','SHA256SUMS','artifacts.json','footprint.json'):
    path=output/name
    assert path.is_file(), f"Missing {path}"
    print(f"{name}: {path.stat().st_size:,} bytes")
```
